## De dónde venimos y qué añadimos hoy



-   En la sesión 2: vectores, producto escalar, norma (longitud).
-   Hoy añadimos la otra magnitud geométrica: el *ángulo*.
-   Y su caso estrella para nosotros: la *ortogonalidad* (ángulo recto).
-   Programa:
    1.  El coseno del ángulo: $\cos\theta = \frac{\mathbf{a}\cdot\mathbf{b}}{\|\mathbf{a}\|\|\mathbf{b}\|}$.
    2.  Ortogonalidad: $\mathbf{a}\cdot\mathbf{b} = 0$.
    3.  *Teorema de Pitágoras* en $\mathbb{R}^n$ (primera demostración seria).
    4.  Cauchy–Schwarz y desigualdad triangular.
-   Método: deducir en $\mathbb{R}^2/\mathbb{R}^3$ (con dibujo), extender a $\mathbb{R}^n$ por analogía.



#### Repetición de la cabecera de la primera transparencia como arranque de los apuntes que la acompañan



En la sesión 2 introdujimos los vectores de $\mathbb{R}^n$ como listas ordenadas de números reales, y dos operaciones sobre ellos: el *producto escalar* $\mathbf{x}\cdot\mathbf{y} = \sum_i x_i y_i$ y la *norma* $\|\mathbf{x}\| = \sqrt{\mathbf{x}\cdot\mathbf{x}}$. Con esas dos herramientas medimos ya una cosa: la *longitud* de un vector.

Hoy añadimos la otra magnitud geométrica fundamental: el *ángulo* entre dos vectores. Y con el ángulo aparece su caso particular más importante para el curso, la *ortogonalidad* (ángulo recto), que es la piedra angular de toda la regresión. Demostraremos el *teorema de Pitágoras* en $\mathbb{R}^n$ —la primera demostración seria del curso— y veremos un par de desigualdades (Cauchy–Schwarz y la triangular) que garantizan que todo el aparato geométrico es coherente.

El plan es honesto en su método: deduciremos las fórmulas en $\mathbb{R}^2$ y $\mathbb{R}^3$, donde podemos dibujar, y las *extenderemos por analogía* a $\mathbb{R}^n$ como definiciones. Esa extensión no es un capricho: es la única que preserva, en cualquier dimensión, las propiedades que ya conocemos en dimensiones bajas.

Todo el contenido de esta sesión está desarrollado, con más detalle y figuras, en el capítulo 11 de mi *Curso de Álgebra Lineal*: [libro online](https://mbujosab.github.io/CursoDeAlgebraLineal/). Os remitiré a él en los puntos donde convenga ver los detalles completos.
