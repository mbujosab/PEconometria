#### De dónde venimos: Pitágoras, otra vez



La lección 4 terminó con una identidad que conviene tener muy presente: el vector de datos $\boldsymbol{y}$ se descompone en dos componentes ortogonales, su proyección sobre $\mathcal{L}(\boldsymbol{1})$ (el vector de medias $\boldsymbol{\mathop{\overline{y}}}=\mu_{\boldsymbol{y}}\boldsymbol{1}$) y su vector de diferencias $\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}$ (perpendicular a $\boldsymbol{1}$). El teorema de Pitágoras, aplicado a esa descomposición, dio
$$
\|\boldsymbol{y}\|_s^2 = \mu_{\boldsymbol{y}}^2 + \sigma_{\boldsymbol{y}}^2.
$$

Hoy vamos a llevar esa misma geometría un paso más allá, en dos direcciones. Primero, vamos a mirar con más detalle la propia varianza $\sigma_{\boldsymbol{y}}^2$ (el cuadrado de la norma), porque de ahí sale una fórmula de cálculo muy conocida en estadística descriptiva. Segundo —y esto es lo verdaderamente nuevo—, vamos a dejar de trabajar con un único vector y vamos a comparar *dos* vectores de datos, $\boldsymbol{x}$ e $\boldsymbol{y}$. Cuando tenemos dos vectores, dos preguntas geométricas naturales son: ¿cuánto vale el producto escalar de sus vectores de diferencias? y ¿qué ángulo forman? Las respuestas son, respectivamente, la *covarianza* y la *correlación*.

Antes de empezar, una cuestión de notación. En la lección 4 introdujimos el producto escalar estadístico $\langle\boldsymbol{x},\boldsymbol{y}\rangle_s=\frac1n(\boldsymbol{x}\cdot\boldsymbol{y})$ y explicamos que hereda de $\boldsymbol{x}\cdot\boldsymbol{y}$ las tres propiedades abstractas (simetría, linealidad, positividad) que bastan para que sigan siendo válidos Pitágoras, Cauchy–Schwarz y la desigualdad triangular. A partir de ahora, y salvo que sea necesario contrastarlo explícitamente con el producto escalar euclídeo, omitiremos el subíndice $s$: escribiremos $\langle\boldsymbol{x},\boldsymbol{y}\rangle$ y $\|\boldsymbol{x}\|$ en todo lo que sigue, dando por hecho que estamos en el contexto estadístico que fijamos la lección pasada.

## La varianza: el cuadrado de la norma



<div class="org-center">
<p>
<i>Definición</i>. La <i>varianza</i> de $\boldsymbol{y}$ es el cuadrado de la norma del vector $\;\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}$:
</p>
</div>

$$
\|\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}\|^2 \; = \;
\sigma_{\boldsymbol{y}}^2 \; = \;
\frac{1}{n}\sum_{i=1}^n (y_i-\mu_{\boldsymbol{y}})^2.
$$



No es una definición nueva: es, literalmente, elevar al cuadrado la norma $\sigma_{\boldsymbol{y}}$ que ya conocíamos de la lección 4.



La identidad de Pitágoras, reescrita despejando:
$$
\|\boldsymbol{y}\|^2 = \mu_{\boldsymbol{y}}^2 + \sigma_{\boldsymbol{y}}^2 \;\Longrightarrow\; \sigma_{\boldsymbol{y}}^2 = \|\boldsymbol{y}\|^2 - \mu_{\boldsymbol{y}}^2.
$$

<div class="org-center">
<p>
¿Qué es exactamente $\|\boldsymbol{y}\|^2$ en términos de sumatorios? Merece la pena mirarlo con calma, con ayuda de un dibujo.
</p>
</div>



#### La varianza: el cuadrado de la norma



En la lección 4 llamamos $\sigma_{\boldsymbol{y}}$ a la *norma* del vector de diferencias $\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}$. La *varianza* no es un objeto distinto: es simplemente el cuadrado de esa norma,
$$
\sigma_{\boldsymbol{y}}^2 = \langle\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}},\,\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}\rangle = \frac{1}{n}\sum_{i=1}^n(y_i-\mu_{\boldsymbol{y}})^2.
$$
Esta es la definición de varianza de cualquier manual de estadística descriptiva. Aquí insistimos en que no aparece de la nada: es el cuadrado de una norma, y esa norma es la que mide la distancia de $\boldsymbol{y}$ al vector constante más próximo, $\boldsymbol{\mathop{\overline{y}}}$.

La identidad de Pitágoras de la lección 4, $\|\boldsymbol{y}\|^2=\mu_{\boldsymbol{y}}^2+\sigma_{\boldsymbol{y}}^2$, puede despejarse para obtener
$$
\sigma_{\boldsymbol{y}}^2 = \|\boldsymbol{y}\|^2 - \mu_{\boldsymbol{y}}^2.
$$
Esta forma alternativa de escribir la varianza —la que en los cursos de estadística descriptiva se suele presentar como "fórmula de cálculo" o "fórmula abreviada"— no es un truco algebraico aislado: es exactamente el teorema de Pitágoras despejado. Para entenderla del todo, solo falta saber qué es $\|\boldsymbol{y}\|^2$ en sumatorios, y eso es justamente lo que hace la siguiente transparencia, apoyándonos en un dibujo.



## Varianza y Pitágoras



Los tres lados del triángulo rectángulo, elevados al cuadrado:
$$
\underbrace{\|\boldsymbol{x}\|^2}_{\text{hipotenusa}^2} \;=\; \underbrace{\|\mu_{\boldsymbol{x}}\boldsymbol{1}\|^2}_{\text{cateto}^2} \;+\; \underbrace{\sigma_{\boldsymbol{x}}^2}_{\text{cateto}^2}.
$$



Como $\|\boldsymbol{1}\|=1$: $\quad\|\mu_{\boldsymbol{x}}\boldsymbol{1}\| = |\mu_{\boldsymbol{x}}|\cdot\|\boldsymbol{1}\| = |\mu_{\boldsymbol{x}}| \;\Longrightarrow\; \|\mu_{\boldsymbol{x}}\boldsymbol{1}\|^2=\mu_{\boldsymbol{x}}^2$.



Despejando:
$$
\boxed{\;\sigma_{\boldsymbol{x}}^2 \;=\; \|\boldsymbol{x}\|^2 - \mu_{\boldsymbol{x}}^2 \;=\; \frac{1}{n}\sum_{i=1}^n x_i^2 \,-\, \mu_{\boldsymbol{x}}^2 \;=\;
\mu_{\boldsymbol{x}^2}  \,-\, \mu_{\boldsymbol{x}}^2;\;}
$$
si denotamos con $\boldsymbol{x}^2$ al vector cuyas componentes son el cuadrado de las de $\boldsymbol{x}$.

<div class="org-center">
<p>
<i>La "fórmula de cálculo" de la varianza no es un truco: es Pitágoras despejado.</i>
</p>
</div>



#### Varianza y Pitágoras



La figura dibuja, sobre cada lado del triángulo rectángulo de la proyección, el cuadrado correspondiente. El área del cuadrado de la hipotenusa es $\|\boldsymbol{x}\|^2$; el área del cuadrado del cateto que va por la recta $\mathcal{L}(\boldsymbol{1})$ es $\|\mu_{\boldsymbol{x}}\boldsymbol{1}\|^2$; y el área del cuadrado del cateto perpendicular es $\sigma_{\boldsymbol{x}}^2$. El teorema de Pitágoras dice que el área grande es la suma de las dos pequeñas.

Falta justificar un paso, que conviene hacer con todo detalle porque es la clave de la fórmula: ¿por qué $\|\mu_{\boldsymbol{x}}\boldsymbol{1}\|^2$ es simplemente $\mu_{\boldsymbol{x}}^2$? La razón es la *homogeneidad de la norma* (vista en la lección 2): para cualquier escalar $\lambda$ y vector $\boldsymbol{v}$, $\|\lambda\boldsymbol{v}\|=|\lambda|\cdot\|\boldsymbol{v}\|$. Aplicándolo con $\lambda=\mu_{\boldsymbol{x}}$ y $\boldsymbol{v}=\boldsymbol{1}$:
$$
\|\mu_{\boldsymbol{x}}\boldsymbol{1}\| = |\mu_{\boldsymbol{x}}|\cdot\|\boldsymbol{1}\|.
$$
Y como fijamos en la lección 4 que $\|\boldsymbol{1}\|=1$ (esa fue precisamente la razón de introducir el factor $\frac1n$), tenemos $\|\mu_{\boldsymbol{x}}\boldsymbol{1}\| = |\mu_{\boldsymbol{x}}|$, y al elevar al cuadrado, $\|\mu_{\boldsymbol{x}}\boldsymbol{1}\|^2 = \mu_{\boldsymbol{x}}^2$. Sin este paso —que solo funciona porque $\|\boldsymbol{1}\|=1$— la fórmula de cálculo no tendría esta forma tan simple.

Sustituyendo en Pitágoras, $\|\boldsymbol{x}\|^2 = \mu_{\boldsymbol{x}}^2 + \sigma_{\boldsymbol{x}}^2$, y despejando:
$$
\sigma_{\boldsymbol{x}}^2 = \|\boldsymbol{x}\|^2 - \mu_{\boldsymbol{x}}^2 = \frac{1}{n}\sum_{i=1}^n x_i^2 - \mu_{\boldsymbol{x}}^2.
$$

*Comprobación con el vector de la figura*. El vector dibujado es $\boldsymbol{x}=(1,5)$. Su media es $\mu_{\boldsymbol{x}}=\frac{1+5}{2}=3$. Por un lado, directamente:
$$
\sigma_{\boldsymbol{x}}^2 = \frac{1}{2}\big[(1-3)^2+(5-3)^2\big] = \frac{1}{2}(4+4) = 4.
$$
Por otro, con la fórmula de cálculo:
$$
\|\boldsymbol{x}\|^2 = \frac{1}{2}(1^2+5^2) = \frac{26}{2}=13, \qquad \sigma_{\boldsymbol{x}}^2 = 13 - 3^2 = 13-9 = 4.
$$
Ambos caminos coinciden, como debían: son la misma cosa vista desde dos ángulos distintos (uno directo, otro pasando por Pitágoras).

Este es un buen momento para insistir en la diferencia de estatus entre las dos fórmulas. $\sigma_{\boldsymbol{x}}^2=\frac1n\sum(x_i-\mu_{\boldsymbol{x}})^2$ es la *definición*: mide directamente la dispersión respecto a la media. $\sigma_{\boldsymbol{x}}^2=\frac1n\sum x_i^2-\mu_{\boldsymbol{x}}^2$ es una *consecuencia* geométrica, útil para el cálculo a mano (solo hace falta $\sum x_i$ y $\sum x_i^2$, sin calcular cada diferencia), pero que oculta el porqué. Con la geometría en la mano, la fórmula de cálculo deja de ser un truco memorizado: es, sencillamente, Pitágoras.

*Para profundizar*:

-   Strang, G. *Introduction to Linear Algebra* (2016), sección sobre varianza y desviación típica como norma en el capítulo dedicado a estadística.



## De un vector a dos: la covarianza



Dados $\boldsymbol{x},\boldsymbol{y}\in\mathbb{R}^n$, consideremos sus *vectores de diferencias* (cada uno respecto a su propia media): $\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}$ e $\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}$.

<div class="org-center">
<p>
<i>Definición</i>. La <i>covarianza</i> entre $\boldsymbol{x}$ e $\boldsymbol{y}$ es el producto escalar de sus vectores de diferencias:
</p>
</div>

$$
\sigma_{\boldsymbol{x}\boldsymbol{y}} = \langle \boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}},\, \boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}\rangle = \frac{1}{n}\sum_{i=1}^n (x_i-\mu_{\boldsymbol{x}})(y_i-\mu_{\boldsymbol{y}}).
$$

-   Si $\boldsymbol{y}=\boldsymbol{x}$: $\;\sigma_{\boldsymbol{x}\boldsymbol{x}} = \sigma_{\boldsymbol{x}}^2$ (la varianza es la covarianza de un vector consigo mismo).
-   *Simetría*: $\sigma_{\boldsymbol{x}\boldsymbol{y}} = \sigma_{\boldsymbol{y}\boldsymbol{x}}$.
-   *Linealidad* en cada argumento.

<div class="org-center">
<p>
Todo heredado del producto escalar: no hay nada nuevo que demostrar.
</p>
</div>



#### De un vector a dos: la covarianza



Hasta ahora trabajábamos con un único vector de datos $\boldsymbol{y}$ y su descomposición en vector de medias y vector de diferencias. La novedad de hoy es tener *dos* vectores de datos, $\boldsymbol{x}$ e $\boldsymbol{y}$, cada uno con su propia media $\mu_{\boldsymbol{x}}$, $\mu_{\boldsymbol{y}}$ y su propio vector de diferencias $\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}$, $\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}$. Ambos vectores de diferencias viven en el mismo subespacio, $\mathcal{L}(\boldsymbol{1})^\perp$ (el conjunto de todos los vectores ortogonales a $\boldsymbol{1}$), así que tiene sentido preguntarse cómo se relacionan entre sí dentro de ese subespacio.

La pregunta más simple que podemos hacer sobre dos vectores es cuánto vale su producto escalar. Esa pregunta, aplicada a los vectores de diferencias, tiene nombre propio en estadística: la *covarianza*.

> *Definición*. La covarianza entre $\boldsymbol{x}$ e $\boldsymbol{y}$ es
> $$
> \sigma_{\boldsymbol{x}\boldsymbol{y}} = \langle\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}},\,\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}\rangle = \frac{1}{n}\sum_{i=1}^n(x_i-\mu_{\boldsymbol{x}})(y_i-\mu_{\boldsymbol{y}}).
> $$Un caso particular reconforta: si tomamos $\boldsymbol{y}=\boldsymbol{x}$, la covarianza se convierte en $\sigma_{\boldsymbol{x}\boldsymbol{x}}=\langle\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}},\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}\rangle=\|\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}\|^2=\sigma_{\boldsymbol{x}}^2$. Es decir, *la varianza es la covarianza de un vector consigo mismo*. Esto no es una coincidencia: la varianza es, geométricamente, el caso degenerado de la covarianza en el que comparamos un vector con él mismo. Todo lo que digamos sobre la covarianza se aplicará, como caso particular, a la varianza.

Como la covarianza es (por definición) un producto escalar, hereda automáticamente sus propiedades: es *simétrica* ($\sigma_{\boldsymbol{x}\boldsymbol{y}}=\sigma_{\boldsymbol{y}\boldsymbol{x}}$, porque $\langle\boldsymbol{a},\boldsymbol{b}\rangle=\langle\boldsymbol{b},\boldsymbol{a}\rangle$) y *lineal* en cada uno de sus dos argumentos. No hace falta demostrar nada nuevo: son las mismas propiedades del producto escalar que fijamos en la lección 2, aplicadas ahora a los vectores de diferencias en lugar de a los vectores originales.

Una advertencia importante, que retomaremos en la siguiente transparencia: la covarianza, por sí sola, no tiene una escala interpretable. Su valor numérico depende de las unidades en que midamos $\boldsymbol{x}$ e $\boldsymbol{y}$ (si $\boldsymbol{x}$ está en euros y lo pasamos a céntimos, la covarianza se multiplica por 100, aunque la relación entre las dos variables no haya cambiado en absoluto). Esto es exactamente lo mismo que le ocurre al producto escalar ordinario: por sí solo no nos dice nada sobre el *ángulo* entre dos vectores, salvo que lo dividamos por sus normas. Ese es el papel que va a jugar la correlación.



## La correlación: el coseno del ángulo



<div class="org-center">
<p>
<i>Definición</i>. La <i>correlación</i> entre $\boldsymbol{x}$ e $\boldsymbol{y}$ es el coseno del ángulo $\theta$ que forman sus vectores de diferencias:
</p>
</div>

$$
\rho_{\boldsymbol{x}\boldsymbol{y}} = \cos\theta = \frac{\langle \boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}},\, \boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}\rangle}{\|\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}\|\,\|\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}\|} = \frac{\sigma_{\boldsymbol{x}\boldsymbol{y}}}{\sigma_{\boldsymbol{x}}\,\sigma_{\boldsymbol{y}}}.
$$



Por Cauchy–Schwarz (lección 3): $\quad -1\leq\rho_{\boldsymbol{x}\boldsymbol{y}}\leq1$.

-   $\rho_{\boldsymbol{x}\boldsymbol{y}}=\pm1$: vectores de diferencias *alineados* ($\theta=0°$ o $180°$).
-   $\rho_{\boldsymbol{x}\boldsymbol{y}}=0$: vectores de diferencias *ortogonales* ($\theta=90°$).

<div class="org-center">
<p>
A diferencia de la covarianza, la correlación no depende de las unidades: es un número entre $[-1,1]$.
</p>
</div>



#### La correlación: el coseno del ángulo



En la lección 3 definimos el coseno del ángulo entre dos vectores cualesquiera como $\cos\theta=\frac{\boldsymbol{a}\cdot\boldsymbol{b}}{\|\boldsymbol{a}\|\|\boldsymbol{b}\|}$, y demostramos que la desigualdad de Cauchy–Schwarz garantiza que ese cociente siempre cae en $[-1,1]$. No hay nada que impida aplicar exactamente la misma definición a los vectores de diferencias $\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}$ e $\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}$: ambos son vectores de $\mathbb{R}^n$ como cualquier otro, y todo lo demostrado en la lección 3 (Pitágoras, Cauchy–Schwarz, la desigualdad triangular) sigue siendo válido, porque se apoya únicamente en las propiedades abstractas del producto escalar, no en cuáles sean concretamente los vectores.

> *Definición*. La correlación entre $\boldsymbol{x}$ e $\boldsymbol{y}$ es
> $$
> \rho_{\boldsymbol{x}\boldsymbol{y}} = \frac{\langle\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}},\,\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}\rangle}{\|\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}\|\,\|\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}\|} = \frac{\sigma_{\boldsymbol{x}\boldsymbol{y}}}{\sigma_{\boldsymbol{x}}\sigma_{\boldsymbol{y}}}.
> $$Es, literalmente, el coseno del ángulo $\theta$ que forman los dos vectores de diferencias. La desigualdad de Cauchy–Schwarz —demostrada con detalle en la lección 3, no simplemente enunciada— nos garantiza que $-1\leq\rho_{\boldsymbol{x}\boldsymbol{y}}\leq1$ siempre, para cualesquiera $\boldsymbol{x}$ e $\boldsymbol{y}$.

Los dos casos extremos son especialmente informativos:

-   $\rho_{\boldsymbol{x}\boldsymbol{y}}=\pm1$ ocurre exactamente cuando los vectores de diferencias están *alineados* (uno es múltiplo escalar no nulo del otro): recordemos que en la lección 3 vimos que la igualdad en Cauchy–Schwarz se da si y solo si los vectores son proporcionales.
-   $\rho_{\boldsymbol{x}\boldsymbol{y}}=0$ ocurre exactamente cuando los vectores de diferencias son *ortogonales*. Este es el sentido geométrico preciso de "no estar correlacionados": no que $\boldsymbol{x}$ e $\boldsymbol{y}$ no tengan relación alguna, sino que sus desviaciones respecto a sus medias son perpendiculares en $\mathbb{R}^n$.

Nótese la diferencia de estatus con la covarianza. La covarianza cambia de valor si cambiamos las unidades de medida (multiplicar $\boldsymbol{x}$ por una constante multiplica $\sigma_{\boldsymbol{x}\boldsymbol{y}}$ por esa misma constante, como veremos en unas transparencias). La correlación, al ser un coseno, es adimensional: no depende de las unidades. Esta es la razón por la que la correlación —y no la covarianza— es la medida habitual para comparar la fuerza de la relación *lineal* entre dos variables.

*Para profundizar*:

-   Wooldridge, J. *Introductory Econometrics* (2020), Apéndice C.3: covarianza y correlación como medidas de asociación lineal.

#### Ilustración geométrica de la correlación



La figura retoma la escena de la lección 4 (dos vectores en $\mathbb{R}^3$ proyectados sobre $\mathcal{L}(\boldsymbol{1})$), pero añade el elemento nuevo de hoy: el ángulo $\theta$ entre los dos vectores de diferencias $\boldsymbol{v}-\boldsymbol{\mathop{\overline{v}}}$ y $\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}$, ambos situados en el plano $\mathcal{L}(\boldsymbol{1})^\perp$ (perpendicular al eje azul). Ese ángulo es precisamente el que determina, a través de su coseno, la correlación $\rho_{\boldsymbol{v}\boldsymbol{x}}$.

La primera imagen muestra la escena en su disposición "natural" en $\mathbb{R}^3$; en ella el ángulo $\theta$ es difícil de apreciar a simple vista, porque los dos vectores de diferencias no están en el plano de la hoja. La segunda imagen es exactamente la misma escena, pero vista desde un punto de observación situado sobre la propia dirección ortogonal a ambos vectores de diferencias: desde ahí, el plano $\mathcal{L}(\boldsymbol{1})^\perp$ se ve "de frente", y el ángulo $\theta$ se aprecia con su verdadera magnitud.

Esta segunda vista anticipa una idea importante que desarrollamos en la siguiente transparencia: para que dos vectores de diferencias puedan formar un ángulo genuino (distinto de $0°$ o $180°$), necesitamos que $\mathcal{L}(\boldsymbol{1})^\perp$ tenga al menos dos dimensiones, es decir, necesitamos $n\geq3$.



## Un hecho geométrico: en $\mathbb{R}^2$ la correlación es siempre $\pm1$



En $\mathbb{R}^2$, $\mathcal{L}(\boldsymbol{1})^\perp$ tiene dimensión $2-1=1$: es una *recta*.

<div class="org-center">
<p>
Todo vector de diferencias en $\mathbb{R}^2$ vive en <i>esa misma recta</i>.
</p>
</div>



Por tanto, dos vectores de diferencias cualesquiera en $\mathbb{R}^2$ están *siempre alineados*:
$$
\rho_{\boldsymbol{x}\boldsymbol{y}} = \pm1 \qquad (\text{en } \mathbb{R}^2).
$$



*Comprobación*, con los ejemplos de la lección 4: $\boldsymbol{x}=(2,4)$ y $\boldsymbol{z}=(2,-6)$ dan $\rho_{\boldsymbol{x}\boldsymbol{z}}=-1$.

<div class="org-center">
<p>
Para ver ángulos <i>genuinos</i> hace falta $n\geq3$. Por eso las figuras de hoy viven en $\mathbb{R}^3$.
</p>
</div>



#### Un hecho geométrico: en $\mathbb{R}^2$ la correlación es siempre $\pm1$



Aquí hay una consecuencia geométrica que suele sorprender la primera vez que se ve, y que conviene detenerse a entender bien, porque explica por qué las figuras de esta lección no podían dibujarse en $\mathbb{R}^2$ (como sí hicimos en la lección 4 para media y desviación típica).

El subespacio $\mathcal{L}(\boldsymbol{1})^\perp$ —el conjunto de todos los vectores ortogonales a $\boldsymbol{1}$, es decir, el conjunto de todos los posibles vectores de diferencias— tiene dimensión $n-1$ dentro de $\mathbb{R}^n$ (porque $\mathcal{L}(\boldsymbol{1})$ ya "ocupa" una dimensión). En $\mathbb{R}^2$, eso significa que $\mathcal{L}(\boldsymbol{1})^\perp$ tiene dimensión $2-1=1$: es una simple *recta* que pasa por el origen, perpendicular a $\boldsymbol{1}$.

Ahora bien: en una recta, *todos* los vectores no nulos son múltiplos escalares unos de otros (dos vectores en una recta están, por definición, alineados). Así que, para cualesquiera dos vectores $\boldsymbol{x},\boldsymbol{y}\in\mathbb{R}^2$ (ninguno de ellos constante), sus vectores de diferencias $\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}$ e $\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}$ viven forzosamente en esa misma recta $\mathcal{L}(\boldsymbol{1})^\perp$, y por tanto están *siempre alineados*. Recordando la lección 3: vectores alineados dan $\cos\theta=\pm1$. Es decir:

> Con solo dos observaciones ($n=2$), la correlación entre dos variables cualesquiera es siempre $\rho=+1$ o $\rho=-1$ (salvo que alguna tenga varianza cero, en cuyo caso la correlación ni siquiera está definida).Verifiquémoslo con dos de los ejemplos numéricos de la lección 4. Tomamos $\boldsymbol{x}=(2,4)$ (con $\mu_{\boldsymbol{x}}=3$, deviaciones $(-1,1)$, $\sigma_{\boldsymbol{x}}=1$) y $\boldsymbol{z}=(2,-6)$ (con $\mu_{\boldsymbol{z}}=-2$, desviaciones $(4,-4)$, $\sigma_{\boldsymbol{z}}=4$). La covarianza:
$$
\sigma_{\boldsymbol{x}\boldsymbol{z}} = \frac{1}{2}\big[(-1)(4)+(1)(-4)\big] = \frac{1}{2}(-8) = -4.
$$
Y la correlación:
$$
\rho_{\boldsymbol{x}\boldsymbol{z}} = \frac{-4}{1\cdot4} = -1.
$$
Efectivamente, $-1$: los vectores de diferencias $(-1,1)$ y $(4,-4)$ son múltiplos el uno del otro (de hecho, $(4,-4)=-4\cdot(-1,1)$), así que están alineados, pero con sentidos opuestos.

Este hecho no es una peculiaridad de estos dos vectores concretos: es forzoso en $\mathbb{R}^2$, sea cual sea el par de vectores que elijamos. La consecuencia práctica es importante: para ilustrar con un dibujo una correlación *distinta* de $\pm1$ (es decir, un ángulo $\theta$ genuino, ni nulo ni llano) necesitamos, como mínimo, $\mathbb{R}^3$ ($n=3$ observaciones), porque solo entonces $\mathcal{L}(\boldsymbol{1})^\perp$ tiene dimensión $2$ y puede albergar de verdad dos direcciones distintas. Por eso las figuras de esta lección —y el ejemplo numérico que viene a continuación— se sitúan en $\mathbb{R}^3$.



## Propiedades geométricas: traslación y escala



-   *Invarianza ante traslaciones*: sumar una constante a $\boldsymbol{x}$ o a $\boldsymbol{y}$ *no* cambia $\sigma_{\boldsymbol{x}\boldsymbol{y}}$ ni $\rho_{\boldsymbol{x}\boldsymbol{y}}$.

-   *Homogeneidad ante escala*: si $\boldsymbol{x}'=\lambda\boldsymbol{x}$ con $\lambda\neq0$:

$$
\sigma_{\boldsymbol{x}'\boldsymbol{y}} = \lambda\,\sigma_{\boldsymbol{x}\boldsymbol{y}}, \qquad \rho_{\boldsymbol{x}'\boldsymbol{y}} = \frac{\lambda}{|\lambda|}\,\rho_{\boldsymbol{x}\boldsymbol{y}} = \pm\rho_{\boldsymbol{x}\boldsymbol{y}}.
$$

<div class="org-center">
<p>
La correlación es invariante a cambios de escala (salvo, quizá, el signo); la covarianza no lo es.
</p>
</div>



#### Propiedades geométricas: traslación y escala



Estas dos propiedades son la versión, para dos vectores, de las que ya vimos en la lección 4 para la desviación típica de un solo vector.

*Invarianza ante traslaciones*. Si $\boldsymbol{x}'=\boldsymbol{x}+a\boldsymbol{1}$ (sumamos una constante $a$ a todas las componentes de $\boldsymbol{x}$), entonces $\mu_{\boldsymbol{x}'}=\mu_{\boldsymbol{x}}+a$, pero el vector de diferencias no cambia:
$$
\boldsymbol{x}'-\boldsymbol{\mathop{\overline{x'}}} = (\boldsymbol{x}+a\boldsymbol{1}) - (\mu_{\boldsymbol{x}}+a)\boldsymbol{1} = \boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}.
$$
Como la covarianza y la correlación solo dependen de los vectores de diferencias, ni $\sigma_{\boldsymbol{x}'\boldsymbol{y}}$ ni $\rho_{\boldsymbol{x}'\boldsymbol{y}}$ cambian. Geométricamente: sumar $a\boldsymbol{1}$ desplaza el vector a lo largo de $\mathcal{L}(\boldsymbol{1})$, sin tocar su componente en $\mathcal{L}(\boldsymbol{1})^\perp$.

*Homogeneidad ante escala*. Si $\boldsymbol{x}'=\lambda\boldsymbol{x}$ con $\lambda\neq0$, entonces $\mu_{\boldsymbol{x}'}=\lambda\mu_{\boldsymbol{x}}$ y el vector de diferencias se escala también por $\lambda$:
$$
\boldsymbol{x}'-\boldsymbol{\mathop{\overline{x'}}} = \lambda\boldsymbol{x}-\lambda\mu_{\boldsymbol{x}}\boldsymbol{1} = \lambda(\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}).
$$
Por la linealidad del producto escalar,
$$
\sigma_{\boldsymbol{x}'\boldsymbol{y}} = \langle\lambda(\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}),\,\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}\rangle = \lambda\,\sigma_{\boldsymbol{x}\boldsymbol{y}}.
$$
Y como $\|\lambda(\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}})\| = |\lambda|\,\|\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}\|$ (homogeneidad de la norma, lección 2), la correlación se transforma como
$$
\rho_{\boldsymbol{x}'\boldsymbol{y}} = \frac{\lambda\,\sigma_{\boldsymbol{x}\boldsymbol{y}}}{|\lambda|\,\sigma_{\boldsymbol{x}}\sigma_{\boldsymbol{y}}} = \frac{\lambda}{|\lambda|}\,\rho_{\boldsymbol{x}\boldsymbol{y}}.
$$
Si $\lambda>0$, $\frac{\lambda}{|\lambda|}=1$ y la correlación no cambia. Si $\lambda<0$, $\frac{\lambda}{|\lambda|}=-1$ y la correlación cambia de signo, pero conserva su magnitud $|\rho_{\boldsymbol{x}\boldsymbol{y}}|$.

La lectura práctica: la covarianza cambia con las unidades (si $\boldsymbol{x}$ pasa de metros a centímetros, $\sigma_{\boldsymbol{x}\boldsymbol{y}}$ se multiplica por 100), pero la correlación —al ser un coseno— es invariante a cambios de escala (salvo, si la escala es negativa, un cambio de signo). Esta es la razón geométrica de por qué la correlación, y no la covarianza, es la medida de asociación lineal que se puede comparar entre variables medidas en unidades distintas.



## Ejemplo numérico completo



$\boldsymbol{x}=(1,2,3)$, $\boldsymbol{y}=(2,2,5)$ (en $\mathbb{R}^3$).

-   $\mu_{\boldsymbol{x}}=2$, $\quad\mu_{\boldsymbol{y}}=3$.
-   Diferencias: $\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}}=(-1,0,1)$, $\quad\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}=(-1,-1,2)$.
-   $\sigma_{\boldsymbol{x}\boldsymbol{y}} = \frac{1}{3}\big[(-1)(-1)+(0)(-1)+(1)(2)\big] = 1$.
-   $\sigma_{\boldsymbol{x}}^2=\frac{2}{3}$, $\quad\sigma_{\boldsymbol{y}}^2=2$.



$$
\rho_{\boldsymbol{x}\boldsymbol{y}} = \frac{1}{\sqrt{\frac{2}{3}\cdot2}} = \frac{1}{\sqrt{4/3}} = \frac{\sqrt{3}}{2} \approx 0{,}866.
$$

<p>
Un ángulo $\theta=30°$: ni ortogonales, ni alineados. Un caso genuino, <i>imposible</i> en $\mathbb{R}^2$.
</p>
</div>



#### Ejemplo numérico completo



Verifiquemos todo lo dicho con un ejemplo completo. Necesitamos $n\geq3$ para que la correlación pueda tomar un valor distinto de $\pm1$, así que trabajamos en $\mathbb{R}^3$: $\boldsymbol{x}=(1,2,3)$, $\boldsymbol{y}=(2,2,5)$.

*Medias*:
$$
\mu_{\boldsymbol{x}} = \frac{1+2+3}{3} = 2, \qquad \mu_{\boldsymbol{y}} = \frac{2+2+5}{3} = 3.
$$

*Vectores de diferencias*:
$$
\boldsymbol{x}-\boldsymbol{\mathop{\overline{x}}} = (1-2,\,2-2,\,3-2) = (-1,0,1), \qquad \boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}} = (2-3,\,2-3,\,5-3) = (-1,-1,2).
$$

*Covarianza*:
$$
\sigma_{\boldsymbol{x}\boldsymbol{y}} = \frac{1}{3}\big[(-1)(-1)+(0)(-1)+(1)(2)\big] = \frac{1}{3}(1+0+2) = 1.
$$

*Varianzas*:
$$
\sigma_{\boldsymbol{x}}^2 = \frac{1}{3}\big[(-1)^2+0^2+1^2\big] = \frac{2}{3}, \qquad \sigma_{\boldsymbol{y}}^2 = \frac{1}{3}\big[(-1)^2+(-1)^2+2^2\big] = \frac{6}{3} = 2.
$$

*Correlación*:
$$
\rho_{\boldsymbol{x}\boldsymbol{y}} = \frac{\sigma_{\boldsymbol{x}\boldsymbol{y}}}{\sigma_{\boldsymbol{x}}\sigma_{\boldsymbol{y}}} = \frac{1}{\sqrt{\frac{2}{3}}\cdot\sqrt{2}} = \frac{1}{\sqrt{\frac{4}{3}}} = \frac{\sqrt{3}}{2} \approx 0{,}866.
$$

Comprobamos que se cumple Cauchy–Schwarz con desigualdad *estricta*: $\sigma_{\boldsymbol{x}\boldsymbol{y}}^2=1 \leq \sigma_{\boldsymbol{x}}^2\sigma_{\boldsymbol{y}}^2=\frac{2}{3}\cdot2=\frac{4}{3}$, y en efecto $1<\frac{4}{3}$, señal de que los vectores de diferencias *no* están alineados (si lo estuvieran, tendríamos igualdad, y $\rho=\pm1$).

El valor $\rho_{\boldsymbol{x}\boldsymbol{y}}=\frac{\sqrt3}{2}$ corresponde a un ángulo $\theta=30°$ entre los vectores de diferencias $(-1,0,1)$ y $(-1,-1,2)$: ni ortogonales ($\theta\neq90°$, $\rho\neq0$), ni alineados ($\theta\neq0°,180°$, $\rho\neq\pm1$). Es exactamente el tipo de situación que, como vimos en la transparencia anterior, es *imposible* de obtener con solo $n=2$ observaciones, y que por eso hemos tenido que buscar en $\mathbb{R}^3$.

*El principio unificador*: la covarianza es un producto escalar; la correlación, un coseno. Nada de esto es nuevo: es la geometría de las lecciones 2 y 3, aplicada ahora a los vectores de diferencias.



#### Recapitulación



El mensaje central de esta sesión es que, al pasar de un vector a dos, no necesitamos ningún concepto nuevo: la covarianza y la correlación son, respectivamente, el producto escalar y el coseno del ángulo, aplicados a los vectores de diferencias. Todo lo que sabíamos sobre productos escalares y ángulos (lecciones 2 y 3) se traslada sin cambios.

Tres consecuencias prácticas:

1.  La varianza es el caso particular $\sigma_{\boldsymbol{x}\boldsymbol{x}}=\sigma_{\boldsymbol{x}}^2$ de la covarianza: comparar un vector consigo mismo.

2.  La desigualdad de Cauchy–Schwarz, demostrada en la lección 3, es la que garantiza $-1\leq\rho_{\boldsymbol{x}\boldsymbol{y}}\leq1$: no hace falta ninguna demostración nueva para justificar por qué la correlación nunca se sale de ese intervalo.

3.  La dimensión de $\mathcal{L}(\boldsymbol{1})^\perp$ importa: en $\mathbb{R}^2$ es una recta, y por eso la correlación entre cualquier par de vectores en $\mathbb{R}^2$ es forzosamente $\pm1$. Se necesita $n\geq3$ para observar correlaciones "genuinas".

La lección 6 llevará este mismo aparato —proyecciones, ortogonalidad, ángulos— a un problema distinto: en lugar de comparar $\boldsymbol{x}$ e $\boldsymbol{y}$, buscaremos la combinación lineal de $\boldsymbol{1}$ y $\boldsymbol{x}$ que mejor aproxima a $\boldsymbol{y}$. Ese será el problema de la regresión lineal simple.

#### Guiño a la sesión siguiente



Hoy hemos visto que comparar dos vectores no exige ningún concepto nuevo: la covarianza es un producto escalar (entre vectores de diferencias) y la correlación es un coseno (entre esos mismos vectores). La próxima sesión es un laboratorio: en él calcularemos, con datos reales, medias, varianzas, covarianzas y correlaciones, y dibujaremos diagramas de dispersión; y comprobaremos, con datos simulados, que las identidades de hoy (en particular, la fórmula de cálculo de la varianza y la relación $\rho=\cos\theta$) se cumplen exactamente.

La lección 6 dará el siguiente gran paso del curso: la *regresión lineal simple*. Hasta ahora hemos proyectado un único vector $\boldsymbol{y}$ sobre la recta $\mathcal{L}(\boldsymbol{1})$ (lección 4); ahora añadiremos un segundo regresor $\boldsymbol{x}$ y proyectaremos sobre el *plano* $\mathcal{L}(\boldsymbol{1},\boldsymbol{x})$ (el conjunto de todas las combinaciones lineales de $\boldsymbol{1}$ y $\boldsymbol{x}$). La lógica —imponer condiciones de ortogonalidad— será exactamente la misma que ya conocemos; solo cambia el subespacio sobre el que proyectamos. Y al final de ese camino, cuando preguntemos "¿cómo de bueno es el ajuste?", la respuesta será, otra vez, un coseno al cuadrado: $R^2=\cos^2\theta$, donde $\theta$ es el ángulo entre el vector centrado $\boldsymbol{y}-\boldsymbol{\mathop{\overline{y}}}$ y su ajuste centrado $\boldsymbol{\mathop{\widehat{y}}}-\boldsymbol{\mathop{\overline{y}}}$. El ángulo de hoy no desaparece: reaparece con otro nombre.

Por último, una pincelada de notación que solo cobrará su forma completa en la lección 12: cuando tengamos varios regresores (por ejemplo, $\boldsymbol{1}$ y $\boldsymbol{x}$), resultará cómodo organizarlos como las columnas de una única matriz $\boldsymbol{X}=(\boldsymbol{1}\;\;\boldsymbol{x})$. El producto $\boldsymbol{X}\boldsymbol{\beta}$ (con $\boldsymbol{\beta}=(\beta_0,\beta_1)$) admite entonces una doble lectura, que usaremos constantemente a partir de ese momento: por un lado, como la lista de los productos escalares de $\boldsymbol{\beta}$ con cada *fila* de $\boldsymbol{X}$; por otro, como la *combinación lineal* $\beta_0\boldsymbol{1}+\beta_1\boldsymbol{x}$ de las *columnas* de $\boldsymbol{X}$. Hoy solo sembramos la idea; la notación matricial completa esperará a que de verdad la necesitemos.
